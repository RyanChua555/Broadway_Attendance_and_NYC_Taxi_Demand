This notebook download the data files from the TLC website and the Broadway data files from kaggle, referencing the code in Tutorial 1.

Import libraries

```python
from urllib.request import urlretrieve
```

Make the paths to put the data in.

```python
import os

output_relative_dir = '../data/'

# check if exists already
if not os.path.exists(output_relative_dir):
    os.makedirs(output_relative_dir)
    
# creat path for each type of data
for target_dir in ('yellow_taxi_data', 'fhvhv_data', 'broadway_data'):
    if not os.path.exists(output_relative_dir + target_dir):
        os.makedirs(output_relative_dir + target_dir)
```

Download all the data files.

```python
YEAR = '2024'
MONTHS = range(1, 13)
```

```python
URL_TEMPLATE_FHVHV = "https://d37ci6vzurychx.cloudfront.net/trip-data/fhvhv_tripdata_"
URL_TEMPLATE_YELLOW = "https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_"
```

```python
yellow_taxi_output_dir = output_relative_dir + 'yellow_taxi_data'
fhvhv_output_dir = output_relative_dir + 'fhvhv_data'
broadway_output_dir = output_relative_dir + 'broadway_data'


for month in MONTHS:
    month = str(month).zfill(2) 
    print(f"Begin month {month}")
    
    # generate url
    url_yellow = f'{URL_TEMPLATE_YELLOW}{YEAR}-{month}.parquet'
    # generate output location and filename
    output_dir = f"{yellow_taxi_output_dir}/{YEAR}-{month}.parquet"
    # download
    urlretrieve(url_yellow, output_dir) 
    
    print(f"Completed month {month}, saved to {output_dir}")
    
    # generate url
    url_fhvhv = f'{URL_TEMPLATE_FHVHV}{YEAR}-{month}.parquet'
    # generate output location and filename
    output_dir = f"{fhvhv_output_dir}/{YEAR}-{month}.parquet"
    # download
    urlretrieve(url_fhvhv, output_dir) 
    
    print(f"Completed month {month}, saved to {output_dir}")
```
